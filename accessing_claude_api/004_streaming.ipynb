```python
# Initial client setup
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Helper functions
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    params={
        "model": model,
        "messages": messages,
        "max_tokens": 1000,
        "temperature": temperature,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)

    return message.content[0].text
```

```python
# Start with an empty message list
messages = []

# Add the user's message to the conversation history
add_user_message(messages, "Give me a one sentence movie idea")
    
# Get Claude's response
answer = chat(messages, temperature=1.0)
answer
```

```python
messages = []

# Add the user's message to the conversation history
add_user_message(messages, "Write me a 1 sentence discription of a fake database")

stream = client.messages.create(
    model=model,
    messages=messages,
    max_tokens=1000,
    stream=True,
)

for event in stream:
    print(event)

```

```python
# Start with an empty message list
messages = []

# Add the user's message to the conversation history
add_user_message(messages, "Write me a 1 sentence discription of a fake database")

with client.messages.stream(
    model=model,
    messages=messages,
    max_tokens=1000,
) as stream:
    for text in stream.text_stream:
        print(text, end="", flush=True)

# Collect the final message and optionally save it
final_message = stream.get_final_message()
print("\n\nFinal message:", final_message)
```
